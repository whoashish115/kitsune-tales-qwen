# Kitsune Tales playground

Runs `Kitsune-Tales-E4B-JP` (Japanese) and `Kitsune-Tales-E4B-EN` (English with Japanese anime themes) as 4-bit GGUF
files through llama.cpp, with the full Gradio playground: genres, title, format, passage, every sampling setting,
length against the requested range, the exact prompt, and download.

**Speed.** On a free T4 GPU (Runtime > Change runtime type > T4 GPU) a short story takes well under a minute; on the free
CPU runtime it takes several minutes. The first run downloads a 5.3 GB model per language.

Site: https://kitsune-tales-qwen.vercel.app · Code: https://github.com/whoashish115/kitsune-tales-qwen
